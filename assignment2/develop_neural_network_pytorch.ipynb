```python
import matplotlib.pyplot as plt
from models.neural_net_pytorch import NeuralNet

input_size = 2
hidden_sizes = [32, 32, 16]
output_size = 3
num_layers = 4
learning_rate = 0.01

net = NeuralNet(input_size, hidden_sizes, output_size, num_layers, learning_rate)

X = torch.randn((8, input_size))
y = torch.randn((8, output_size))

losses = []
for epoch in range(100):
    loss_val = net.backward(X, y)
    losses.append(loss_val)

plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.show()
```
